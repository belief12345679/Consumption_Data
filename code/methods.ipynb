# Part 1: Global Coordination Level (GCL)
> **Source:** Levy, O., Amit, G., Vaknin, D. et al. *Age-related loss of gene-to-gene transcriptional coordination among single cells*. Nat Metab 2 (2020).

## 1. Concept Overview
[cite_start]**Global Coordination Level (GCL)** is a "top-down" statistical measure designed to quantify the systemic interdependence (coordination) within a high-dimensional system[cite: 54].

[cite_start]Unlike traditional correlation analyses (like Pearson) that look at pairs of variables (e.g., Gene A vs. Gene B)[cite: 35], GCL measures the **overall dependency** of the entire system. It answers the question: *"To what extent does the state of a random subset of variables predict the state of the remaining variables?"*

In the context of **consumer data**, GCL can be used to measure the "structuredness" of consumption patterns. A high GCL implies strong budget constraints or rigid lifestyle patterns, while a low GCL implies stochastic or uncoordinated spending behavior.

## 2. The Algorithm
[cite_start]The calculation of GCL involves a bootstrapping procedure to estimate the multivariate dependency of the system[cite: 55, 62]:

1.  [cite_start]**Random Split:** Randomly divide the $N$ variables (e.g., genes or product categories) into two non-overlapping subsets, $S_1$ and $S_2$[cite: 56].
2.  [cite_start]**Calculate Dependency:** Measure the multivariate dependency between these two subsets using **Bias-Corrected Distance Correlation (bcdCorr)**[cite: 59].
    * [cite_start]*Why Distance Correlation?* Unlike Pearson correlation, it captures non-linear relationships and is zero if and only if the variables are independent[cite: 101, 102].
3.  [cite_start]**Bootstrap:** Repeat steps 1 and 2 for $m$ iterations (typically $m=50$ is sufficient for stabilization)[cite: 549].
4.  [cite_start]**Average:** The GCL is defined as the average dependency level across all iterations[cite: 62].

## 3. Mathematical Formulation
Given a data matrix $X$ (Variables $\times$ Samples), the GCL is calculated as:

$$
GCL(X) = \frac{1}{m} \sum_{k=1}^{m} \text{bcdCorr}(X_{S_1^{(k)}}, X_{S_2^{(k)}})
$$

Where:
* [cite_start]$m$ is the number of bootstrap iterations[cite: 548].
* [cite_start]$S_1^{(k)}$ and $S_2^{(k)}$ are the random variable subsets for iteration $k$[cite: 545].
* [cite_start]$\text{bcdCorr}$ is the bias-corrected distance correlation[cite: 518].

# Part 2: The AIDS Model (Almost Ideal Demand System)
> **Source:** Deaton, A., & Muellbauer, J. (1980). *An Almost Ideal Demand System*. American Economic Review.

## 1. Concept Overview
The **Almost Ideal Demand System (AIDS)** is a seminal model in microeconomics used to analyze how consumers allocate a fixed budget across different commodities. It is widely used for estimating price and income elasticities.

While GCL measures the *structure* of the data, AIDS models the *mechanisms* of allocation (e.g., *"If the price of Rice goes up by 1%, how does the share of spending on Wheat change?"*).

## 2. The Core Equation
The model estimates the **Expenditure Share** ($w_i$) of a specific good $i$. The share is a function of prices ($p$) and real income (represented by total expenditure $X$).

$$
w_i = \alpha_i + \sum_{j} \gamma_{ij} \ln(p_j) + \beta_i \ln\left(\frac{X}{P}\right) + \varepsilon_i
$$

Where:
* $w_i$: The share of total budget spent on good $i$ (e.g., Spending on Coffee / Total Budget).
* $p_j$: The price of good $j$.
* $X$: Total expenditure (budget).
* $P$: An aggregate price index (deflator).
* $\varepsilon_i$: Error term.

## 3. Interpreting Parameters
When you run a regression on this model, the coefficients tell you the following:

* **$\beta_i$ (Income Parameter):**
    * If $\beta_i > 0$: The good is a **Luxury** (spending share increases as you get richer).
    * If $\beta_i < 0$: The good is a **Necessity** (spending share decreases as you get richer).
* **$\gamma_{ij}$ (Price Parameter):**
    * Measures the change in the share of good $i$ with respect to a change in the price of good $j$.
    * Used to calculate **Cross-Price Elasticity** (substitutes vs. complements).

## Summary Comparison

| Feature | GCL (Global Coordination Level) | AIDS (Almost Ideal Demand System) |
| :--- | :--- | :--- |
| **Goal** | Quantify system-wide order/chaos | Estimate elasticity & budget allocation |

| **Key Insight** | "How *tightly coupled* are consumer choices?" | "How does *price/income* change choice?" |
| **Best For** | Exploratory Analysis, Anomaly Detection | Policy Analysis, Pricing Strategy |

# Part 3:

# Methodology: Filtering Systemic Shocks via Random Matrix Theory (RMT)

## 1. Introduction to Random Matrix Theory in Time-Series Analysis

In the analysis of complex systems, such as consumer expenditure or human mobility, data is inherently noisy. Random Matrix Theory (RMT) provides a rigorous mathematical framework to filter out this stochastic noise and identify true, underlying deterministic signals (such as macroeconomic shocks or collective synchronization). 

Originally developed by Eugene Wigner and Freeman Dyson to describe the energy levels of heavy atomic nuclei, RMT compares the empirical properties of a system's cross-correlation matrix against a theoretical "null hypothesis" of pure random noise, defined by the Marchenko-Pastur Law. Deviations from this law indicate the presence of systemic shocks.

---

## 2. Mathematical Formulation

### 2.1. Data Normalization
Consider a multivariate time-series dataset of $N$ consumer expenditure items, observed over a time period of length $T$. Let $X_i(t)$ be the expenditure of item $i$ at time $t$. To ensure stationarity, we first calculate the log-returns of the expenditure:

$$G_i(t) = \ln X_i(t + \Delta t) - \ln X_i(t)$$

Next, we normalize the time-series to have a mean of zero and a variance of one:

$$x_i(t) = \frac{G_i(t) - \langle G_i \rangle}{\sigma_i}$$

Where $\langle G_i \rangle$ is the temporal average and $\sigma_i$ is the standard deviation.

### 2.2. The Empirical Cross-Correlation Matrix
We construct the $N \times N$ empirical cross-correlation matrix, denoted as $\mathbf{C}$. The elements $C_{ij}$ represent the Pearson correlation coefficient between items $i$ and $j$:

$$C_{ij} = \frac{1}{T} \sum_{t=1}^{T} x_i(t) x_j(t)$$

By definition, the diagonal elements $C_{ii} = 1$, and $C_{ij} \in [-1, 1]$.

### 2.3. Eigenvalue Decomposition
To analyze the structural properties of $\mathbf{C}$, we solve the characteristic equation to find its eigenvalues $\lambda_k$ and eigenvectors $\mathbf{u}_k$:

$$\mathbf{C} \mathbf{u}_k = \lambda_k \mathbf{u}_k$$

This yields a set of $N$ eigenvalues, ordered such that $\lambda_1 \le \lambda_2 \le \dots \le \lambda_N$.

---

## 3. The Null Hypothesis: Marchenko-Pastur Law

To determine which eigenvalues represent true economic signals and which represent noise, we compare the empirical eigenvalue distribution to the **Marchenko-Pastur (MP) distribution**. The MP law defines the theoretical spectrum of eigenvalues for a purely random correlation matrix (Wishart matrix) of the same dimensions $N$ and $T$.

As $N, T \rightarrow \infty$ with the aspect ratio $Q = T/N \ge 1$ held constant, the probability density function of the eigenvalues is given by:

$$P_{MP}(\lambda) = \frac{Q}{2\pi \sigma^2 \lambda} \sqrt{(\lambda_{max} - \lambda)(\lambda - \lambda_{min})}$$

Where the upper and lower bounds of the noise spectrum are defined as:

$$\lambda_{max}^{min} = \sigma^2 \left( 1 \pm \sqrt{\frac{1}{Q}} \right)^2$$



---

## 4. Identification of Potential Shocks

The application of the MP Law allows us to partition the empirical eigenvalues into two regimes:

* **The Random Noise Regime ($\lambda_{min} \le \lambda_k \le \lambda_{max}$):** The bulk of the eigenvalues fall within the theoretical MP bounds. These correspond to independent, idiosyncratic fluctuations in individual expenditure items that contain no systemic information.
* **The Deviating/Signal Regime ($\lambda_k > \lambda_{max}$):** Eigenvalues that fall strictly above the theoretical maximum bound cannot be explained by chance. These "deviating eigenvalues" represent collective synchronization phenomena.

### 4.1. The "Giant Eigenvalue" ($\lambda_{max}$)
In economic time-series, the largest eigenvalue typically separates significantly from the bulk. This "Giant Eigenvalue" acts as the **Market Mode**. The corresponding eigenvector contains positive weights for almost all expenditure items. 

When an exogenous shock (e.g., COVID-19) occurs, the magnitude of this Giant Eigenvalue increases dramatically, mathematically signifying that the system has synchronized and the diverse expenditure items are now moving collectively in response to the shock.
